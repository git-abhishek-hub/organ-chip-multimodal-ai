```python
meta_clean = meta.drop_duplicates(
    subset="Metadata_well_id"
)

cp_clean = cp.drop_duplicates(
    subset="Metadata_well_id"
)
```

```python
meta_clean.shape
```

```python
cp_clean.shape
```

```python
merged_clean = meta_clean.merge(
    cp_clean,
    on="Metadata_well_id",
    how="inner"
)
```

```python
merged_clean.shape
```

```python
feature_cols = [
    c for c in merged_clean.columns
    if not c.startswith("Metadata_")
]

variances = merged_clean[feature_cols].var()

low_var_features = variances[
    variances == 0
].index

len(low_var_features)
```

```python
merged_clean = merged_clean.drop(
    columns=low_var_features
)
merged_clean.shape
```

```python
merged_clean.shape
```

```python
import pycytominer as pycytominer
from pycytominer.feature_select import feature_select

cp_filtered = feature_select(
    merged_clean,
    features=feature_cols,
    operation=[
        "drop_na_columns",
        "variance_threshold",
        "correlation_threshold"
    ],
    corr_threshold=0.9
)

cp_filtered.shape
```

Making Master Table.

```python
common_cols = [
    c for c in cp_filtered.columns
    if c.startswith("Metadata_")
]
len(common_cols)
```

```python
morph_features = [
    c for c in cp_filtered.columns
    if not c.startswith("Metadata_")
]

len(morph_features)
```

```python
master_table = best_dose.merge(
    cp_filtered[
        ["Metadata_well_id"] + morph_features
    ],
    on="Metadata_well_id",
    how="left"
)

master_table.shape
```
